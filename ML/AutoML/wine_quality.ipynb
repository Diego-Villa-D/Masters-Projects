```python
import pandas as pd, h2o
from h2o.automl import H2OAutoML
```

```python
# OPCION 1 - Clasificación multiclase

# --- DECISION HUMANA: elegir el dataset y la columna objetivo ---
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"

df = pd.read_csv(url, sep=';')

objetivo = "quality"

# Iniciar H2O
h2o.init()

# Convertir DataFrame de pandas a H2OFrame
datos = h2o.H2OFrame(df)

# Variables predictoras
x = datos.columns
x.remove(objetivo)

# --- DECISION HUMANA: tratar quality como clasificación multiclase ---
datos[objetivo] = datos[objetivo].asfactor()

# División entrenamiento/prueba
train, test = datos.split_frame(ratios=[0.8], seed=42)

# --- DECISION HUMANA: métrica de éxito y presupuesto de tiempo ---
aml = H2OAutoML(
    max_runtime_secs=180,
    seed=42,
    sort_metric="logloss"      # Métrica adecuada para clasificación multiclase
)

# Entrenamiento
aml.train(
    x=x,
    y=objetivo,
    training_frame=train
)

# Leaderboard
print(aml.leaderboard.head())

# Evaluación en el conjunto de prueba
perf = aml.leader.model_performance(test)
print(perf)
```

```python
# --- DECISION HUMANA: elegir el dataset y la columna objetivo ---
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"

df = pd.read_csv(url, sep=';')


df["high_q"] = (df["quality"] >= 7).astype(int)

objetivo = "high_q"

# Iniciar H2O
h2o.init()

# Convertir DataFrame de pandas a H2OFrame
datos = h2o.H2OFrame(df)

# Variables predictoras
x = datos.columns
x.remove(objetivo)
x.remove("quality")

# --- DECISION HUMANA: tratar quality como clasificación binaria ---
datos[objetivo] = datos[objetivo].asfactor()

# División entrenamiento/prueba
train, test = datos.split_frame(ratios=[0.8], seed=42)

# --- DECISION HUMANA: métrica de éxito y presupuesto de tiempo ---
aml = H2OAutoML(
    max_runtime_secs=180,
    seed=42,
    sort_metric="AUC"      # Métrica adecuada para clasificación binaria
)

# Entrenamiento
aml.train(
    x=x,
    y=objetivo,
    training_frame=train
)

# Leaderboard
print(aml.leaderboard.head())

# Evaluación en el conjunto de prueba
perf = aml.leader.model_performance(test)
print(perf)
```
