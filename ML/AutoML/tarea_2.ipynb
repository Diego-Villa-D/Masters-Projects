```python
import pandas as pd, h2o
from h2o.automl import H2OAutoML
```

```python
df = pd.read_csv("titanic.csv")                 # Cargar dataset   
df.head()
```

```python
objetivo = "Survived"
df = df.drop(columns=["PassengerId", "Name", "Ticket", "Cabin"]) # limpieza humana

h2o.init()

datos = h2o.H2OFrame(df)

x = datos.columns; x.remove(objetivo)

datos[objetivo] = datos[objetivo].asfactor() # DECISION HUMANA: es clasificacion
train, test = datos.split_frame([0.8], seed=42)

# --- DECISION HUMANA: metrica de exito y presupuesto de tiempo ---
aml = H2OAutoML(max_runtime_secs=180, seed=42, sort_metric="AUC")
aml.train(x=x, y=objetivo, training_frame=train)
print(aml.leaderboard.head()) # modelos que probo
print("AUC:", aml.leader.model_performance(test).auc())
```
